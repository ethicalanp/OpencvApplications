```python
import cv2
import matplotlib.pyplot as plt
```

```python
#load image
image = cv2.imread('../data/test_images/Ronaldo.jpg')

image_rgb = cv2.cvtColor(
    image,
    cv2.COLOR_BGR2RGB
)

plt.figure(figsize=(8,6))
plt.imshow(image_rgb)
plt.title("Orginal Image")
plt.axis("off")
```

```python
#Convert to Grayscale

gray=cv2.cvtColor(
    image,
    cv2.COLOR_BGR2GRAY
)

plt.figure(figsize=(8,6))
plt.imshow(gray,cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")
```

```python
#Apply Medium Blur 

gray_blur=cv2.medianBlur(
    gray,
    5
)
plt.figure(figsize=(8,6))
plt.imshow(gray_blur,cmap='gray')
plt.title("Median Blurred Image")
plt.axis("off")
```

```python
#Detect Edges 

edges = cv2.adaptiveThreshold(
    gray_blur,
    255,
    cv2.ADAPTIVE_THRESH_MEAN_C,
    cv2.THRESH_BINARY,
    9,
    9
)

plt.figure(figsize=(8,6))
plt.imshow(edges,cmap='gray')
plt.title("Detected Edges")
plt.axis("off")
```

```python
#Smooth Colors

color = cv2.bilateralFilter(
    image_rgb,
    9,
    300,
    300
)

plt.figure(figsize=(8,6))
plt.imshow(color)
plt.title("Smoothed Colors")
plt.axis("off")
```

```python
#Combine edge Mask with Colors

cartoon = cv2.bitwise_and(
    color,
    color,
    mask=edges
)

plt.figure(figsize=(8,6))
plt.imshow(cartoon)
plt.title("Cartoon Image")
plt.axis("off")
```

```python
#comparing 
plt.figure(figsize=(14,7))
plt.subplot(1,2,1)
plt.imshow(image_rgb)
plt.title("Original")
plt.axis("off")
plt.subplot(1,2,2)
plt.imshow(cartoon)
plt.title("Cartoon Filter")
plt.axis("off")
plt.show()
```

```python
#Save output
cartoon_bgr = cv2.cvtColor(
    cartoon,
    cv2.COLOR_RGB2BGR
)
cv2.imwrite(
    "../outputs",
    cartoon_bgr
)
print("Cartoon image saved successfully!")
```
